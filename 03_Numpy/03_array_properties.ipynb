```python
import numpy as np
arr = np.array([[1,2,3],
               [4,5,6]])
print(arr)
```

```python
print(arr.ndim) #no of dimension
```

```python
print(arr.shape) #print rows and columns numbers
```

```python
print(arr.size) #total no of elements
```

```python
print(arr.dtype) #data type of elements
```

```python
print(arr.itemsize) #bytes used by one element
```

```python
print(arr.nbytes) #total memory used by elements
```

```python
#type casting

float_arr = arr.astype(np.float64)
print(type(float_arr))
print(float_arr.dtype)
```
